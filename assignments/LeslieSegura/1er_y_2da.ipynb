```python
# La funcion "FOR" va a evaluar sobre cada posición de la secuencia
def ContenidoGC_v2(DNA):
  counter=0
  for base in DNA:
    if base=="C" or base=="G":
      counter= counter + 1
  return counter/len(DNA)
```
